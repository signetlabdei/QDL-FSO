```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os

legend_titles = {
    "qdl_fso_lowtran_slant_to_space.csv": "Zenith angle",
    "qdl_fso_lowtran_desert_wind.csv": "Wind speed",
    "qdl_fso_lowtran_atmospheric_model_no_aerosol.csv": "Atmospheric model",
    "qdl_fso_lowtran_vertical_volcanic.csv": "Volcanic type and extinction profile",
    "qdl_fso_lowtran_navy_windspeed_avgwindspeed.csv": "Wind speed - average wind speed",
    "qdl_fso_lowtran_navy_airmasscharacter.csv": "Air mass character",
    "qdl_fso_lowtran_kim_kruse.csv": "Attenuation model and parameters",
    # files not listed here get no legend title
}

for file in glob.glob("qdl_fso_lowtran_*.csv"):
    figure_title = os.path.splitext(file)[0]
    df = pd.read_csv(file)
    x_col = df.columns[0]
    for col in df.columns[1:]:
        plt.scatter(df[x_col], df[col], label=col, s=5)
    plt.ylim(0, 1)
    plt.legend(title=legend_titles.get(file))
    plt.title(figure_title)
    plt.xlabel(x_col)
    plt.ylabel("transmissivity")

    # plt.savefig(f"./figures/{figure_title}.pdf")
    plt.savefig(f"./figures/{figure_title}.png")
    
    plt.show()
```
